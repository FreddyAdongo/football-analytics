```python
# Get games for a specific league and season
nwsl_games = asa_client.get_games(leagues="nwsl", seasons="2026")
```

```python
# Filter by team
nc_games = asa_client.get_games(leagues="nwsl", team_names="North Carolina Courage", seasons=2026)
teams = asa_client.get_teams(leagues="nwsl")
managers = asa_client.get_managers(leagues="nwsl")

# Merge home and away team names
nc_games = nc_games.merge(
    teams[["team_id", "team_name"]].rename(columns={"team_id": "home_team_id", "team_name": "home_team_name"}),
    on="home_team_id", how="left"
).merge(
    teams[["team_id", "team_name"]].rename(columns={"team_id": "away_team_id", "team_name": "away_team_name"}),
    on="away_team_id", how="left"
)

# Merge home and away manager names
nc_games = nc_games.merge(
    managers[["manager_id", "manager_name"]].rename(columns={"manager_id": "home_manager_id", "manager_name": "home_manager_name"}),
    on="home_manager_id", how="left"
).merge(
    managers[["manager_id", "manager_name"]].rename(columns={"manager_id": "away_manager_id", "manager_name": "away_manager_name"}),
    on="away_manager_id", how="left"
)

# Select clean columns
nc_games = nc_games[[
    "game_id", "date_time_utc", "season_name", "matchday",
    "home_team_name", "away_team_name",
    "home_score", "away_score",
    "home_manager_name", "away_manager_name",
    "stadium_id", "attendance", "status"
]]

nc_games
```
